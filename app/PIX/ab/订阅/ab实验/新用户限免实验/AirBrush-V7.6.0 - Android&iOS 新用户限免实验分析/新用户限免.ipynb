```python
data = pd.read_csv('py_0428.csv')#id 维度的收入、save pv
print(data)
```

```python

 # 输入评估指标
index_name = ['sub_success_to_paid_gmv','ads_bookings']
platform_name =['ANDROID']

i=0
n= len(index_name)

j=0
m=len(platform_name)

a_group =[11075]   #对照组
b_group =[11077]  #实验组

while j < m :
    pla = platform_name[j]
    A= data.query("ab_code == @a_group and platform == @pla" )
    B= data.query("ab_code ==  @b_group and platform == @pla" )
    
    index= index_name[i]
    rvs1 = A[index]
    rvs2 = B[index]

    # h0 A>=B H1 A<B
    t_statistic, p_value = stats.ttest_ind(rvs1,rvs2, equal_var=True)
    if p_value <0.05 :
        t ='显著'
    else:
        t = '不显著'

    print(pla,':',index,"t统计量:", t_statistic,"p值:", p_value,t)
    
    i +=1 
```

```python
name = 'ads_bookings'
A= data.query("ab_code ==  11072 or ab_code ==  11075  " )
B= data.query("ab_code == 11073 or ab_code ==  11076 " )
C= data.query("ab_code == 11074 or ab_code ==  11077 " )
rvs1 = A[name]
rvs2 = B[name]
rvs3 = C[name]
# h0 A>=B H1 A<B
t_statistic, p_value = stats.ttest_ind(rvs1,rvs2, equal_var=True)

print("t统计量:", t_statistic)
print("p值:", p_value)


t_statistic, p_value = stats.ttest_ind(rvs1,rvs3, equal_var=True)

print("t统计量:", t_statistic)
print("p值:", p_value)
```

```python
name = 'save_pv'
A= data.query("(ab_code ==  11072 or ab_code ==  11075 ) " )
B= data.query("ab_code == 11073 or ab_code ==  11076 " )
C= data.query("ab_code == 11074 or ab_code ==  11077 " )
 
rvs1 = A[A[name]!= 0][name] # 删除 0 

rvs2 = B[B[name]!= 0][name] 
rvs3 = C[C[name]!= 0][name] 

t_statistic, p_value = stats.ttest_ind(rvs1,rvs2, equal_var=True)

print("t统计量:", t_statistic)
print("p值:", p_value)


t_statistic, p_value = stats.ttest_ind(rvs1,rvs3, equal_var=True)

print("t统计量:", t_statistic)
print("p值:", p_value)

```
